```python
er = 25
n1 = 1
n2 = np.sqrt(1 / er)
gamma = (n2 - n1) / (n2 + n1)
tau = (2*n2) / (n2 + n1)
print(f'a.) Γ = {gamma:.2f}, τ = {tau:.2f}')

S = (1 + abs(gamma)) / (1 - abs(gamma))
print(f'b.) S = {S:.2f}')

E0i = 20

Pi_avg = abs(E0i)**2 / (2*n1) / 377

Pr_avg = abs(gamma)**2 * Pi_avg

Pt_avg = abs(tau)**2 * abs(E0i)**2 / (2 * n2) / 377

print(f'c.) Pi_avg = {Pi_avg:.2f}, Pr_avg = {Pr_avg:.2f}, Pt_avg = {Pt_avg:.2f} W/m^2')
```
